# Bài 1: Giới thiệu Delta Lake & Delta Log

## Mục tiêu
- Hiểu Delta Lake là gì, giải quyết vấn đề gì so với data lake thuần Parquet.
- Hiểu cấu trúc thư mục của một bảng Delta và cơ chế **transaction log** (`_delta_log`).
- Hiểu 4 tính chất **ACID** mà Delta Lake mang lại trên nền object storage.


## 1.1. Delta Lake là gì?

**Delta Lake** là một storage layer mã nguồn mở, xây dựng trên nền các file **Parquet**, bổ sung:

- **ACID transactions**: mọi thao tác ghi (write/update/delete/merge) là nguyên tử — hoặc thành công hoàn toàn, hoặc không có gì thay đổi. Không bao giờ có trạng thái "ghi dở".
- **Schema enforcement & evolution**: bảng có schema rõ ràng, từ chối dữ liệu sai kiểu, nhưng vẫn có thể mở rộng schema một cách có kiểm soát.
- **Time travel**: đọc lại bất kỳ phiên bản nào trong quá khứ của bảng.
- **Unified batch & streaming**: một bảng Delta vừa là sink cho streaming, vừa là nguồn cho batch.
- **Scalable metadata**: metadata (danh sách file, thống kê...) được lưu dưới dạng Spark job có thể xử lý, thay vì phải `list` hàng triệu file trên object storage (vốn rất chậm với S3/MinIO).

### Vấn đề của "data lake" thuần Parquet/Hive

| Vấn đề | Parquet/Hive thuần | Delta Lake |
|---|---|---|
| Ghi đè giữa chừng bị lỗi | Dữ liệu rác, đọc phải file half-written | Transaction log đảm bảo all-or-nothing |
| Nhiều job ghi đồng thời | Race condition, mất dữ liệu | Optimistic concurrency control |
| Đọc trong lúc đang ghi | Có thể đọc phải dữ liệu không nhất quán | Snapshot isolation — luôn đọc 1 version nhất quán |
| Đổi schema | Thủ công, dễ vỡ downstream | Schema enforcement + evolution có kiểm soát |
| "Quay lại" dữ liệu cũ | Không có, phải tự backup | `VERSION AS OF` / `TIMESTAMP AS OF` |

Đây chính là nền tảng của kiến trúc **Lakehouse**: kết hợp chi phí thấp + khả năng mở rộng của data lake với các đảm bảo giao dịch của data warehouse.


## 1.2. Cấu trúc vật lý của một bảng Delta

```
s3a://data-platform/managed/bai01.db/events/
├── _delta_log/
│   ├── 00000000000000000000.json   # commit 0 (CREATE / ghi lần đầu)
│   ├── 00000000000000000001.json   # commit 1
│   ├── 00000000000000000002.json   # commit 2
│   ├── ...
│   └── _last_checkpoint            # (xuất hiện sau khi có checkpoint)
├── part-00000-....snappy.parquet   # file dữ liệu thật
├── part-00001-....snappy.parquet
└── ...
```

- **Dữ liệu thật** luôn là các file **Parquet** chuẩn — Delta không đổi định dạng file dữ liệu.
- **`_delta_log/`** chứa các file JSON — mỗi file là **một transaction (commit)**, đánh số tuần tự tăng dần bắt đầu từ `0`.
- Mỗi commit JSON gồm các "action" như:
  - `add` — thêm 1 file Parquet vào bảng (kèm thống kê: min/max, số dòng, null count...)
  - `remove` — đánh dấu 1 file không còn thuộc bảng (tombstone, **không xoá vật lý ngay**)
  - `metaData` — schema, partition columns, table properties
  - `commitInfo` — ai, khi nào, thao tác gì (INSERT/UPDATE/DELETE/MERGE/OPTIMIZE...)
  - `protocol` — phiên bản reader/writer tối thiểu để đọc/ghi bảng an toàn

**Trạng thái hiện tại của bảng** = tổng hợp (replay) tất cả action từ commit `0` đến commit mới nhất. Đây chính là lý do Delta Lake có thể *time travel*: chỉ cần replay tới một commit cũ hơn.

Sau một số commit (mặc định 10), Spark tự ghi thêm một **checkpoint** (`.checkpoint.parquet`) — một bản snapshot toàn bộ state dưới dạng Parquet, để không phải đọc lại toàn bộ lịch sử JSON mỗi lần mở bảng.


## 1.3. ACID trong Delta Lake

- **Atomicity**: một job ghi tạo ra đúng 1 file JSON commit mới; nếu job crash giữa chừng, file JSON không được ghi → coi như chưa có gì xảy ra.
- **Consistency**: schema và constraints được kiểm tra trước khi commit được chấp nhận.
- **Isolation**: dùng **optimistic concurrency control** — mỗi writer đọc version hiện tại, làm việc độc lập, rồi cố gắng commit version kế tiếp; nếu bị giành mất (conflict), writer đó tự động thử lại hoặc báo lỗi (xem Bài 11).
- **Durability**: một khi file JSON commit đã được ghi thành công lên storage (S3/MinIO), thay đổi là vĩnh viễn.

````python
from pyspark.sql import SparkSession

DB_NAME = "bai01"
WAREHOUSE_DIR = "s3a://data-platform/managed"

spark = (
    SparkSession.builder
    .appName("bai01-gioi-thieu-delta")
    .master("spark://spark-master:7077")
    .config("spark.driver.host", "spark-master")
    .config("spark.sql.warehouse.dir", WAREHOUSE_DIR)
    .enableHiveSupport()
    .getOrCreate()
)

spark.sql(f"CREATE DATABASE IF NOT EXISTS {DB_NAME} LOCATION '{WAREHOUSE_DIR}/{DB_NAME}.db'")
spark.sql(f"USE {DB_NAME}")
spark
````

## 1.4. Ví dụ minh hoạ

Tạo một bảng Delta đơn giản, rồi "mổ xẻ" trực tiếp thư mục `_delta_log` để nhìn thấy transaction log thật.


````python
TABLE_NAME = "events"

spark.sql(f"DROP TABLE IF EXISTS {TABLE_NAME}")

df1 = spark.createDataFrame(
    [(1, "click", "home"), (2, "click", "cart")],
    ["id", "event_type", "page"],
)
df1.write.format("delta").mode("overwrite").saveAsTable(TABLE_NAME)
print("Commit 0: tạo bảng với 2 dòng")
````

````python
# Ghi thêm (append) -> tạo thêm 1 commit mới trong _delta_log
df2 = spark.createDataFrame([(3, "purchase", "checkout")], ["id", "event_type", "page"])
df2.write.format("delta").mode("append").saveAsTable(TABLE_NAME)
print("Commit 1: append thêm 1 dòng")

spark.sql(f"SELECT * FROM {TABLE_NAME} ORDER BY id").show()
````

````python
%load_ext sparksql_magic
````

````python
%%sparksql
DESCRIBE EXTENDED events
````

````python
# Lấy đường dẫn vật lý của bảng trên MinIO rồi liệt kê _delta_log bằng chính Spark (qua Hadoop FS API)
table_location = (
    spark.sql(f"DESCRIBE EXTENDED {TABLE_NAME}")
    .filter("col_name = 'Location'")
    .collect()[0]["data_type"]
)
print("Table location:", table_location)

sc = spark.sparkContext
hadoop_conf = sc._jsc.hadoopConfiguration()
path = sc._jvm.org.apache.hadoop.fs.Path(f"{table_location}/_delta_log")
fs = path.getFileSystem(hadoop_conf)
for f in fs.listStatus(path):
    print(f.getPath().getName(), f.getLen(), "bytes")
````

````python
# Đọc nội dung 1 file commit JSON để thấy action add/metaData/commitInfo
log_path = sc._jvm.org.apache.hadoop.fs.Path(f"{table_location}/_delta_log/00000000000000000000.json")
stream = fs.open(log_path)
content = sc._jvm.org.apache.commons.io.IOUtils.toString(stream, "UTF-8")
stream.close()

import json as _json
for line in content.strip().split("\n"):
    action = _json.loads(line)
    print(list(action.keys()), "->", {k: v for k, v in action.items() if k in ("commitInfo", "metaData")} or action.get("add", {}).get("path"))
````

## Gợi ý / đáp án tham khảo

````python
# Đáp án Bài 1
spark.sql("DROP TABLE IF EXISTS bai01.products")
products = spark.createDataFrame(
    [(1, "Keyboard", 25.5), (2, "Mouse", 12.0), (3, "Monitor", 199.9)],
    ["id", "name", "price"],
)
products.write.format("delta").mode("overwrite").saveAsTable("bai01.products")
spark.sql("SELECT * FROM bai01.products").show()
````

````python
# Đáp án Bài 2
more = spark.createDataFrame([(4, "Webcam", 45.0), (5, "Headset", 30.0)], ["id", "name", "price"])
more.write.format("delta").mode("append").saveAsTable("bai01.products")

loc = spark.sql("DESCRIBE EXTENDED bai01.products").filter("col_name = 'Location'").collect()[0]["data_type"]
log_dir = sc._jvm.org.apache.hadoop.fs.Path(f"{loc}/_delta_log")
fs2 = log_dir.getFileSystem(sc._jsc.hadoopConfiguration())
json_files = [f.getPath().getName() for f in fs2.listStatus(log_dir) if f.getPath().getName().endswith(".json")]
print(sorted(json_files))
print("Số commit:", len(json_files))
````

````python
# Đáp án Bài 3
data_dir = sc._jvm.org.apache.hadoop.fs.Path(loc)
parquet_files = [f.getPath().getName() for f in fs2.listStatus(data_dir) if f.getPath().getName().endswith(".parquet")]
print("Số file Parquet:", len(parquet_files))
# Có thể > số lần write vì mỗi write có thể sinh nhiều file (theo số partition/executor)
````

````python
# Đáp án Bài 4
last_commit_name = sorted(json_files)[-1]
p = sc._jvm.org.apache.hadoop.fs.Path(f"{loc}/_delta_log/{last_commit_name}")
s = fs2.open(p)
txt = sc._jvm.org.apache.commons.io.IOUtils.toString(s, "UTF-8")
s.close()
for line in txt.strip().split("\n"):
    print(_json.loads(line))
````

**Đáp án Bài 5**: Không. Delta Lake ghi dữ liệu vào các file Parquet **mới**, và chỉ khi toàn bộ file dữ liệu đã ghi xong, nó mới ghi **1 file JSON commit duy nhất** vào `_delta_log` (thường bằng thao tác tạo file "put-if-absent"/atomic rename tuỳ storage). Nếu job bị kill trước khi file JSON commit được ghi xong, sẽ có vài file Parquet "mồ côi" nằm trên storage nhưng **không được nhắc tới trong bất kỳ commit nào**, nên khi đọc bảng qua Delta (replay `_delta_log`), các file mồ côi đó hoàn toàn bị bỏ qua — bảng vẫn ở trạng thái nhất quán của lần commit thành công gần nhất. Các file rác này sau đó có thể được dọn bằng `VACUUM` (Bài 8).

# Bài 10: Structured Streaming với Delta Lake

## Mục tiêu
- Dùng bảng Delta làm **streaming source** (`readStream`) và **streaming sink** (`writeStream`).
- Hiểu vai trò của **checkpoint location** và các loại `trigger` (`availableNow`, `processingTime`, `once`).
- Dùng `foreachBatch` để chạy `MERGE` (upsert) trong 1 streaming pipeline — mẫu **stream → Delta upsert** rất phổ biến.


## 10.1. Delta Lake là "unified batch & streaming"

Vì mọi thay đổi trên bảng Delta đều là các **commit tuần tự** trong `_delta_log`, Spark Structured Streaming có thể coi 1 bảng Delta như một **stream vô hạn các append** — mỗi lần đọc, engine chỉ cần biết "đã xử lý tới commit nào" (lưu trong checkpoint) rồi đọc tiếp các commit mới hơn. Điều này biến Delta thành cầu nối tự nhiên giữa batch và streaming: **cùng 1 bảng** vừa nhận ghi từ streaming job, vừa được batch job khác đọc bình thường bằng SQL.

## 10.2. Đọc bảng Delta dạng stream

```python
stream_df = spark.readStream.format("delta").table("db.source_tbl")
```
Theo mặc định, streaming source **chỉ đọc các phép `append`** đã xảy ra kể từ khi query bắt đầu (hoặc từ checkpoint). Nếu source có `UPDATE`/`DELETE`/`overwrite`, mặc định sẽ **báo lỗi** (vì đó không phải "chỉ thêm mới") — trừ khi bật `option("ignoreChanges", "true")` hoặc `option("ignoreDeletes", "true")`, hoặc dùng CDF (`option("readChangeFeed", "true")`, xem Bài 9) để nhận đúng cả update/delete.

## 10.3. Ghi vào bảng Delta dạng stream

```python
(stream_df.writeStream
    .format("delta")
    .option("checkpointLocation", "s3a://.../_checkpoints/my_stream")
    .trigger(availableNow=True)     # xu ly het du lieu dang co roi dung (giong batch job chay dinh ky)
    .toTable("db.sink_tbl"))
```

- **`checkpointLocation`**: nơi Spark lưu tiến trình (offset đã xử lý tới đâu) — **bắt buộc**, và **duy nhất cho mỗi query** (không dùng chung giữa 2 query khác nhau).
- **Trigger**:
  - `trigger(processingTime="10 seconds")` — chạy liên tục theo chu kỳ, phù hợp streaming "thật" 24/7.
  - `trigger(availableNow=True)` — xử lý hết dữ liệu hiện có rồi tự dừng; dùng cho pipeline chạy theo lịch (giống Airflow trigger mỗi giờ) mà vẫn tận dụng cơ chế checkpoint/exactly-once của streaming.
  - `trigger(once=True)` — tương tự `availableNow` nhưng xử lý trong 1 batch duy nhất (không chia nhỏ), đã deprecated, khuyến nghị dùng `availableNow`.

## 10.4. `foreachBatch` — chạy MERGE trong streaming

Muốn **upsert** (không chỉ append) trong streaming sink, dùng `foreachBatch` để áp dụng logic tuỳ ý (kể cả `MERGE`) cho từng micro-batch:

```python
def upsert_batch(batch_df, batch_id):
    (DeltaTable.forName(spark, "db.sink_tbl").alias("t")
        .merge(batch_df.alias("s"), "t.id = s.id")
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute())

(stream_df.writeStream
    .foreachBatch(upsert_batch)
    .option("checkpointLocation", "...")
    .trigger(availableNow=True)
    .start())
```
Đây là mẫu chuẩn cho **CDC ingestion**: đọc thay đổi liên tục (qua CDF hoặc source khác), rồi `MERGE` vào bảng đích thay vì chỉ append.

## 10.5. Ví dụ minh hoạ

````python
# Bang nguon (streaming source) va bang dich (streaming sink)
spark.sql("DROP TABLE IF EXISTS bai10.raw_events")
spark.sql("DROP TABLE IF EXISTS bai10.events_append_sink")

spark.sql("""
CREATE TABLE bai10.raw_events (event_id INT, user_id STRING, amount DOUBLE)
USING DELTA
""")

# Ghi du lieu "batch 1" vao nguon TRUOC khi bat dau stream
spark.createDataFrame([(1,"u1",10.0),(2,"u2",20.0)], ["event_id","user_id","amount"]) \
    .write.format("delta").mode("append").saveAsTable("bai10.raw_events")
````

````python
CHECKPOINT_1 = "s3a://data-platform/managed/bai10.db/_checkpoints/append_sink"

stream_df = spark.readStream.format("delta").table("bai10.raw_events")

query = (
    stream_df.writeStream
    .format("delta")
    .option("checkpointLocation", CHECKPOINT_1)
    .trigger(availableNow=True)     # xu ly het du lieu hien co roi tu dung, khong chay mai mai
    .toTable("bai10.events_append_sink")
)
query.awaitTermination()   # cho query xu ly xong (vi la availableNow nen se tu ket thuc)
print("Stream 1 hoan tat, trang thai:", query.lastProgress)
````

````python
# Them batch 2 vao nguon, chay lai stream voi CUNG checkpoint -> chi xu ly phan MOI (incremental)
spark.createDataFrame([(3,"u3",30.0)], ["event_id","user_id","amount"]) \
    .write.format("delta").mode("append").saveAsTable("bai10.raw_events")

query2 = (
    spark.readStream.format("delta").table("bai10.raw_events")
    .writeStream.format("delta")
    .option("checkpointLocation", CHECKPOINT_1)
    .trigger(availableNow=True)
    .toTable("bai10.events_append_sink")
)
query2.awaitTermination()

spark.sql("SELECT * FROM bai10.events_append_sink ORDER BY event_id").show()
# Ca 3 dong deu co, nhung batch 2 (event_id=3) duoc them qua 1 lan chay stream RIENG, chi doc phan moi
````

````python
# foreachBatch + MERGE: upsert streaming thay vi chi append
from delta.tables import DeltaTable

spark.sql("DROP TABLE IF EXISTS bai10.user_totals")
spark.sql("CREATE TABLE bai10.user_totals (user_id STRING, total_amount DOUBLE) USING DELTA")

def upsert_batch(batch_df, batch_id):
    agg = batch_df.groupBy("user_id").sum("amount").withColumnRenamed("sum(amount)", "total_amount")
    (
        DeltaTable.forName(spark, "bai10.user_totals").alias("t")
        .merge(agg.alias("s"), "t.user_id = s.user_id")
        .whenMatchedUpdate(set={"total_amount": "t.total_amount + s.total_amount"})
        .whenNotMatchedInsertAll()
        .execute()
    )
    print(f"batch {batch_id}: upserted {agg.count()} user(s)")

CHECKPOINT_2 = "s3a://data-platform/managed/bai10.db/_checkpoints/upsert_sink"
query3 = (
    spark.readStream.format("delta").table("bai10.raw_events")
    .writeStream
    .foreachBatch(upsert_batch)
    .option("checkpointLocation", CHECKPOINT_2)
    .trigger(availableNow=True)
    .start()
)
query3.awaitTermination()

spark.sql("SELECT * FROM bai10.user_totals ORDER BY user_id").show()
````

## Gợi ý / đáp án tham khảo

````python
# Dap an Bai 1
spark.sql("DROP TABLE IF EXISTS bai10.raw_orders")
spark.sql("CREATE TABLE bai10.raw_orders (order_id INT, status STRING, amount DOUBLE) USING DELTA")
spark.createDataFrame([(1,"pending",50.0),(2,"pending",75.0)], ["order_id","status","amount"]) \
    .write.format("delta").mode("append").saveAsTable("bai10.raw_orders")
````

````python
# Dap an Bai 2
spark.sql("DROP TABLE IF EXISTS bai10.orders_mirror")
CKPT_MIRROR = "s3a://data-platform/managed/bai10.db/_checkpoints/orders_mirror"
q = (
    spark.readStream.format("delta").table("bai10.raw_orders")
    .writeStream.format("delta")
    .option("checkpointLocation", CKPT_MIRROR)
    .trigger(availableNow=True)
    .toTable("bai10.orders_mirror")
)
q.awaitTermination()
spark.sql("SELECT * FROM bai10.orders_mirror ORDER BY order_id").show()
````

````python
# Dap an Bai 3
spark.createDataFrame([(3,"paid",100.0),(4,"paid",20.0)], ["order_id","status","amount"]) \
    .write.format("delta").mode("append").saveAsTable("bai10.raw_orders")

q2 = (
    spark.readStream.format("delta").table("bai10.raw_orders")
    .writeStream.format("delta")
    .option("checkpointLocation", CKPT_MIRROR)   # CUNG checkpoint nhu Bai 2
    .trigger(availableNow=True)
    .toTable("bai10.orders_mirror")
)
q2.awaitTermination()
spark.sql("SELECT * FROM bai10.orders_mirror ORDER BY order_id").show()
````

````python
# Dap an Bai 4
from delta.tables import DeltaTable
spark.sql("DROP TABLE IF EXISTS bai10.status_counts")
spark.sql("CREATE TABLE bai10.status_counts (status STRING, cnt LONG) USING DELTA")

def upsert_counts(batch_df, batch_id):
    agg = batch_df.groupBy("status").count().withColumnRenamed("count", "cnt")
    (
        DeltaTable.forName(spark, "bai10.status_counts").alias("t")
        .merge(agg.alias("s"), "t.status = s.status")
        .whenMatchedUpdate(set={"cnt": "t.cnt + s.cnt"})
        .whenNotMatchedInsertAll()
        .execute()
    )

CKPT_COUNTS = "s3a://data-platform/managed/bai10.db/_checkpoints/status_counts"
q3 = (
    spark.readStream.format("delta").table("bai10.raw_orders")
    .writeStream.foreachBatch(upsert_counts)
    .option("checkpointLocation", CKPT_COUNTS)
    .trigger(availableNow=True)
    .start()
)
q3.awaitTermination()
spark.sql("SELECT * FROM bai10.status_counts ORDER BY status").show()
````

**Đáp án Bài 5**: Streaming query sẽ **ném lỗi** ngay khi phát hiện source có thay đổi không phải append thuần (thường là `UnsupportedOperationException` / báo "data update/deletion not supported"), vì mặc định Structured Streaming giả định nguồn chỉ tăng trưởng (append-only) để đảm bảo tính đúng đắn incremental. Cách xử lý đúng khi nguồn thật sự cần hỗ trợ update:
1. Bật **Change Data Feed** trên bảng nguồn (Bài 9) và đọc bằng `option("readChangeFeed", "true")` — nhận đúng cả `insert`/`update_preimage`/`update_postimage`/`delete`, rồi tự viết logic `foreachBatch` (thường là `MERGE`) áp dụng đúng ngữ nghĩa từng loại thay đổi vào sink.
2. Nếu chỉ cần "bỏ qua an toàn" mà chấp nhận có thể trễ/không chính xác 100%, dùng `option("ignoreChanges", "true")` (chấp nhận đọc lại cả file bị update, có thể tạo trùng lặp cần dedup ở downstream) hoặc `option("ignoreDeletes", "true")` nếu chỉ có xoá theo partition ở đầu bảng.

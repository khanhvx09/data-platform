# Bài 11: Concurrency Control & Transaction Conflicts

## Mục tiêu
- Hiểu **optimistic concurrency control (OCC)** trong Delta Lake — vì sao không dùng lock truyền thống.
- Nhận diện các loại exception khi 2 transaction xung đột: `ConcurrentAppendException`, `ConcurrentDeleteReadException`...
- Hiểu 2 isolation level: `Serializable` và `WriteSerializable`.


## 11.1. Vì sao Optimistic Concurrency Control?

Object storage (S3/MinIO) không hỗ trợ khoá (lock) đa tiến trình theo kiểu database truyền thống. Delta Lake giải quyết ghi đồng thời bằng **optimistic concurrency control**:

1. Mỗi transaction đọc **version hiện tại** của bảng khi bắt đầu (snapshot isolation khi đọc).
2. Transaction thực hiện công việc của nó (tính toán, ghi file Parquet mới) **hoàn toàn độc lập**, không khoá bảng.
3. Khi commit, transaction cố gắng ghi file `_delta_log/{N+1}.json` — thao tác này dựa vào tính **atomic "put-if-absent"** của storage (chỉ 1 writer có thể tạo thành công file tên `N+1`, các writer khác ghi cùng tên sẽ thất bại).
4. Nếu bị "giành" version `N+1`, transaction thua sẽ **kiểm tra xung đột (conflict check)**: so sánh những gì nó dựa vào (file đã đọc, điều kiện WHERE...) với những gì transaction thắng vừa thay đổi.
   - Nếu **không thực sự xung đột về mặt logic** (ví dụ 2 transaction append vào 2 partition khác nhau) → tự động **rebase** lên version mới và **retry** commit ở version `N+2`.
   - Nếu **có xung đột thật** (cùng đụng vào 1 file/1 dòng dữ liệu) → transaction thua bị huỷ, ném exception lên tầng ứng dụng — ứng dụng phải tự quyết định retry hay không.

## 11.2. Các loại conflict exception thường gặp

| Exception | Ý nghĩa |
|---|---|
| `ConcurrentAppendException` | 2 transaction cùng thêm file vào cùng 1 vùng dữ liệu (ví dụ 2 `MERGE`/`UPDATE` cùng động vào 1 partition) mà 1 trong 2 cần đọc đúng tập file ổn định |
| `ConcurrentDeleteReadException` | Transaction đang đọc 1 file mà transaction khác đã xoá (`remove`) file đó trước |
| `ConcurrentDeleteDeleteException` | 2 transaction cùng cố xoá cùng 1 file |
| `MetadataChangedException` | Schema/table properties bị đổi (vd. `ALTER TABLE`) giữa lúc transaction khác đang chạy |
| `ProtocolChangedException` | Phiên bản reader/writer protocol của bảng bị nâng cấp giữa chừng |
| `ConcurrentTransactionException` | 2 transaction cùng `appId`/`txnVersion` (thường gặp khi streaming source bị chạy trùng) |

Khi gặp các exception này, cách xử lý thông thường ở tầng ứng dụng là **retry toàn bộ transaction** (đọc lại dữ liệu mới nhất rồi thử ghi lại) — Delta cố tình không tự ý retry giúp vì có thể logic tính toán của transaction phụ thuộc vào dữ liệu đã đọc (retry ngầm có thể cho kết quả sai).

## 11.3. Isolation Level

- **`Serializable`** (mặc định cho hầu hết các thao tác: UPDATE/DELETE/MERGE) — mức cô lập chặt nhất, đảm bảo kết quả tương đương chạy các transaction tuần tự.
- **`WriteSerializable`** (mặc định cho **blind append**, tức chỉ ghi thêm không đọc trạng thái hiện tại) — nới lỏng hơn 1 chút để tăng khả năng ghi song song, vẫn đảm bảo an toàn cho trường hợp append đơn thuần.

Cấu hình qua table property: `ALTER TABLE db.tbl SET TBLPROPERTIES ('delta.isolationLevel' = 'Serializable')`.

## 11.4. Ví dụ minh hoạ

Mô phỏng 2 "writer" chạy đồng thời bằng 2 thread Python cùng gọi vào 1 `SparkSession`, cùng `UPDATE` chồng lấn dữ liệu trên 1 bảng — quan sát 1 transaction thắng, 1 transaction bị từ chối.


````python
import concurrent.futures

spark.sql("DROP TABLE IF EXISTS bai11.counters")
spark.sql("CREATE TABLE bai11.counters (id INT, val INT) USING DELTA")
spark.sql("INSERT INTO bai11.counters VALUES (1, 0)")

def bump(worker_name):
    spark.sql(f"UPDATE bai11.counters SET val = val + 1 WHERE id = 1")
    return f"{worker_name}: COMMIT THANH CONG"

with concurrent.futures.ThreadPoolExecutor(max_workers=2) as pool:
    futures = {pool.submit(bump, name): name for name in ["writer-A", "writer-B"]}
    for fut in concurrent.futures.as_completed(futures):
        name = futures[fut]
        try:
            print(fut.result())
        except Exception as e:
            print(f"{name}: THAT BAI - {type(e).__name__}: {str(e)[:150]}")

spark.sql("SELECT * FROM bai11.counters").show()
````

````python
# Xem lich su commit de xac nhan CHI 1 trong 2 UPDATE duoc ghi nhan thanh cong tai 1 thoi diem
# (transaction thua co the tu retry va thanh cong o lan sau, tuy logic ung dung - o vi du tren, code KHONG tu retry
#  nen neu that bai thi mat luon lan update do)
spark.sql("DESCRIBE HISTORY bai11.counters").select("version","operation","operationMetrics").show(truncate=False)
````

````python
# Retry thu cong khi gap conflict exception - mau pattern chuan trong ung dung thuc te
import time

def bump_with_retry(worker_name, max_retries=5):
    for attempt in range(max_retries):
        try:
            spark.sql("UPDATE bai11.counters SET val = val + 1 WHERE id = 1")
            return f"{worker_name}: thanh cong sau {attempt+1} lan thu"
        except Exception as e:
            if attempt == max_retries - 1:
                raise
            time.sleep(0.2)   # backoff ngan roi thu lai voi du lieu/version moi nhat
    return f"{worker_name}: khong thanh cong"

with concurrent.futures.ThreadPoolExecutor(max_workers=2) as pool:
    futures = [pool.submit(bump_with_retry, name) for name in ["writer-C", "writer-D"]]
    for fut in concurrent.futures.as_completed(futures):
        print(fut.result())

spark.sql("SELECT * FROM bai11.counters").show()
````

## Gợi ý / đáp án tham khảo

````python
# Dap an Bai 1
import concurrent.futures
spark.sql("DROP TABLE IF EXISTS bai11.stock")
spark.sql("CREATE TABLE bai11.stock (product_id INT, qty INT) USING DELTA")
spark.sql("INSERT INTO bai11.stock VALUES (1, 100)")

def sell(name):
    spark.sql("UPDATE bai11.stock SET qty = qty - 10 WHERE product_id = 1")
    return f"{name}: OK"

with concurrent.futures.ThreadPoolExecutor(max_workers=2) as pool:
    futs = {pool.submit(sell, n): n for n in ["A","B"]}
    for f in concurrent.futures.as_completed(futs):
        try:
            print(f.result())
        except Exception as e:
            print(futs[f], "FAILED:", type(e).__name__)

spark.sql("SELECT * FROM bai11.stock").show()
````

````python
# Dap an Bai 2
import time
def sell_retry(name, max_retries=5):
    for i in range(max_retries):
        try:
            spark.sql("UPDATE bai11.stock SET qty = qty - 10 WHERE product_id = 1")
            return f"{name}: OK sau {i+1} lan"
        except Exception:
            if i == max_retries - 1:
                raise
            time.sleep(0.2)

with concurrent.futures.ThreadPoolExecutor(max_workers=2) as pool:
    futs = [pool.submit(sell_retry, n) for n in ["C","D"]]
    for f in concurrent.futures.as_completed(futs):
        print(f.result())

spark.sql("SELECT * FROM bai11.stock").show()   # qty giam dung 20 so voi truoc do (10 tu Bai1 * so lan thanh cong + 10*2 o day)
````

````python
# Dap an Bai 3
spark.sql("DROP TABLE IF EXISTS bai11.sales")
spark.sql("CREATE TABLE bai11.sales (id INT, region STRING, amount DOUBLE) USING DELTA PARTITIONED BY (region)")
spark.sql("INSERT INTO bai11.sales VALUES (0, 'INIT', 0.0)")

def append_region(region):
    spark.sql(f"INSERT INTO bai11.sales VALUES (1, '{region}', 100.0)")
    return f"append {region}: OK"

with concurrent.futures.ThreadPoolExecutor(max_workers=2) as pool:
    futs = {pool.submit(append_region, r): r for r in ["north", "south"]}
    for f in concurrent.futures.as_completed(futs):
        try:
            print(f.result())
        except Exception as e:
            print(futs[f], "FAILED:", type(e).__name__)
# Ca 2 thuong deu THANH CONG vi la blind append vao 2 partition khac nhau -> khong xung dot du lieu that
spark.sql("SELECT * FROM bai11.sales").show()
````

**Đáp án Bài 4**:
```sql
ALTER TABLE bai11.stock SET TBLPROPERTIES ('delta.isolationLevel' = 'WriteSerializable');
```
Cân nhắc `WriteSerializable` khi workload chủ yếu là **append thuần** (ví dụ nhiều producer cùng ghi log/event vào 1 bảng, không đọc lại trạng thái hiện tại để quyết định ghi gì) và muốn tối đa hoá throughput ghi song song — chấp nhận 1 mức nới lỏng nhỏ về thứ tự quan sát được (không đảm bảo tuần tự hoá hoàn toàn như `Serializable`) để đổi lấy ít conflict/retry hơn. Với bảng có `UPDATE`/`DELETE`/`MERGE` phụ thuộc vào việc đọc đúng trạng thái hiện tại (như bài toán số dư `stock`/`accounts`), nên giữ `Serializable` mặc định để tránh đọc phải dữ liệu "cũ" giữa các transaction đồng thời.

**Đáp án Bài 5**: Vì Delta không biết **logic nghiệp vụ** đằng sau câu lệnh. Với `UPDATE ... SET price = price * 1.1`, nếu tự động retry mà không đọc lại dữ liệu mới nhất, kết quả có thể áp dụng nhân 1.1 lên **giá trị cũ đã lỗi thời** (dữ liệu gốc đã bị transaction khác đổi entre lúc đó), dẫn tới kết quả sai lệch với ý định người dùng — hoặc tệ hơn, nếu giá trị `price` được tính từ kết quả của 1 query phức tạp trước đó (không chỉ là 1 câu SQL đơn giản), việc âm thầm retry có thể dùng nhầm input đã stale. Vì vậy Delta để tầng ứng dụng **tự quyết định** cách retry đúng đắn (đọc lại dữ liệu mới, tính lại, rồi mới ghi) thay vì đoán hộ.
